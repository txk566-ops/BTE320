```python
import random

play = input("Do you want to play Rock, Paper, Scissors? (yes/no): ")

while play == "yes":

    # user move
    user_move = input("Enter rock, paper, or scissors: ")

    # computer randomly chooses
    computer_move = random.choice(["rock", "paper", "scissors"])

    print("Computer chose:", computer_move)

    if user_move == computer_move:
        print("Tie!")

    elif user_move == "rock":
        if computer_move == "scissors":
            print("You win!")
        else:
            print("Computer wins!")

    elif user_move == "paper":
        if computer_move == "rock":
            print("You win!")
        else:
            print("Computer wins!")

    elif user_move == "scissors":
        if computer_move == "paper":
            print("You win!")
        else:
            print("Computer wins!")

    else:
        print("Invalid input.")

    # ask to play again
    play = input("Do you want to play again? (yes/no): ")

print("Game over.")
```

```python
def checkEmailAddress(email):
    assert isinstance(email, str), "Input must be a string"

    if email == "":
        raise ValueError("Empty input")

    if email.count("@") != 1:
        raise ValueError("No @ symbol or too many @ symbols")

    parts = email.split("@")
    username = parts[0]
    domain_full = parts[1]

    if "." not in domain_full:
        raise ValueError("No . in domain")

    domain_parts = domain_full.split(".")
    domain = domain_parts[0]
    top_domain = domain_parts[1]

    if username == "":
        raise ValueError("Missing username")

    if domain == "":
        raise ValueError("Missing domain")

    if top_domain == "":
        raise ValueError("Missing top-level domain")

    return True


try:
    email = input("Enter email address: ")
    result = checkEmailAddress(email)
    print(result)

except ValueError as e:
    print(e)

except:
    print("Error")
```

```python
def getValues(n):
    assert isinstance(n, int), "n must be an integer"
    if n < 2:
        raise ValueError("List must contain at least 2 values to proceed.")

    values = []
    i = 0
    while i < n:
        try:
            val = float(input(f"  Enter value {i + 1}: "))
            if val < 0:
                raise ValueError("Negative weights are not valid.")
            values.append(val)
            i += 1
        except ValueError as e:
            print(f"  Warning: {e} Please try again.")
    return values


def calcAverage(alist):
    assert isinstance(alist, list), "Input must be a list"
    try:
        return sum(alist) / len(alist)
    except ZeroDivisionError:
        raise ZeroDivisionError("No values remain after excluding the first two.")


try:
    try:
        n = int(input("How many values will you enter? "))
    except ValueError:
        raise ValueError("Must input a number.")
    weights = getValues(n)
    trimmed = weights[2:]
    average = calcAverage(trimmed)
    print(f"Average (excluding first two values): {average:.2f}")
except ValueError as e:
    print(f"Error: {e}")
except ZeroDivisionError as e:
    print(f"Error: {e}")
except AssertionError as e:
    print(f"Assertion failed: {e}")
```

```python
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal):
        if formal:
            print(f"{self.last}, {self.first}")
        else:
            print(self.first)

    def __str__(self):
        return f"{self.first} {self.last}"

first_name = input("Enter first name: ")
last_name = input("Enter last name: ")

person = Greeting(first_name, last_name)


choice = input("Formal display? (yes/no): ").lower()

if choice == "yes":
    person.display(True)
else:
    person.display(False)

print("Full name:", person)
```

```python
class Point(object):
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        x_diff_sq = (self.x - other.x)**2
        y_diff_sq = (self.y - other.y)**2
        return (x_diff_sq + y_diff_sq)**0.5


class Rectangle(object):
    def __init__(self, p1, p2, p3, p4):
        self.p1 = p1
        self.p2 = p2
        self.p3 = p3
        self.p4 = p4
        self.width = p1.distance(p2)
        self.height = p1.distance(p4)

    def calc_width(self):
        return self.p1.distance(self.p2)

    def calc_height(self):
        return self.p1.distance(self.p4)

    def area(self):
        return self.width * self.height



p1 = Point(0, 0)
p2 = Point(4, 0)
p3 = Point(4, 3)
p4 = Point(0, 3)

rect = Rectangle(p1, p2, p3, p4)
print(rect.calc_width())
print(rect.calc_height())
print(rect.area())
```

```python
class Employee:
    def __init__(self, name, hourly_wage, hours_worked):
        self.name = name
        self.__wage = hourly_wage
        self.hours_worked = hours_worked

    def get_wage(self):
        return self.__wage

    def set_wage(self, new_wage):
        if new_wage < 0:
            print("Negative wages are not allowed")
        else:
            self.__wage = new_wage

    def calculate_pay(self):
        return self.__wage * self.hours_worked

    def display_info(self):
        print("Name:", self.name)
        print("Hourly Wage:", self.__wage)
        print("Hours Worked:", self.hours_worked)


class Manager(Employee):
    def __init__(self, name, hourly_wage, hours_worked, bonus):
        super().__init__(name, hourly_wage, hours_worked)
        self.bonus = bonus

    def calculate_pay(self):
        return self.get_wage() * self.hours_worked + self.bonus

    def display_info(self):
        Employee.display_info(self)
        print("Bonus:", self.bonus)


emp = Employee("Alice", 20, 40)
mgr = Manager("Bob", 30, 40, 500)

print("=== Regular Employee ===")
emp.display_info()
print("Weekly Pay:", emp.calculate_pay())

print()

print("=== Manager ===")
mgr.display_info()
print("Weekly Pay:", mgr.calculate_pay())
```
